## 4.

The **Gumbel distribution** is often used to model extreme events in hydraulic engineering, such as river flow peaks.  
A continuous random variable $X$ follows a Gumbel distribution if its probability density function is:

$$
f_X(x) = \frac{1}{\sigma} \exp\left[-\left(\frac{x - \mu}{\sigma} + \exp\left(-\frac{x - \mu}{\sigma}\right)\right)\right], \quad x \in \mathbb{R}
$$

where $\mu \in \mathbb{R}$ is the location parameter and $\sigma \in \mathbb{R}^+$ is the scale parameter.

For flood‑risk management, two critical levels are defined:

- **Alert Level** $L_A$: the flow value exceeded with probability $0.1$
- **Emergency Level** $L_E = L_A + 1000 \,\text{m}^3/\text{s}$

In the Grijalva River (Mexico), the maximum daily flow $X$ (in m³/s) follows a Gumbel distribution with parameters:

- $\mu = 3150$
- $\sigma = 560$

Using the `extraDistr` package, determine:

1. The exact value of  
   $$P(X > L_E \mid X > L_A).$$

2. For a sample of size $n = 2700$ generated with seed **3730**, compute the proportion of observations exceeding $L_E$ among those exceeding $L_A$.

Report the ratio:

$$
\frac{\text{Simulated value}}{\text{Exact value}}
$$

rounded to four decimal places.
